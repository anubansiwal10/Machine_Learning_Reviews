```python
outlier_summary = []

for col in sensor_columns:

    Q1 = X[col].quantile(0.25)
    Q3 = X[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((X[col] < lower) | (X[col] > upper)).sum()

    percentage = (outliers / len(X)) * 100

    outlier_summary.append({
        "Feature": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower,
        "Upper Bound": upper,
        "Outlier Count": outliers,
        "Outlier %": percentage
    })

outlier_df = pd.DataFrame(outlier_summary)

outlier_df = outlier_df.sort_values(
    "Outlier %",
    ascending=False
)

display(outlier_df.round(3))
```

```python
X_hourly = X.copy()

X_hourly["hour"] = X_hourly["date_time"].dt.floor("h")

hourly_mean = X_hourly.groupby("hour")[sensor_columns].mean()
hourly_mean.columns = [
    f"{col}_mean" for col in sensor_columns
]

hourly_std = X_hourly.groupby("hour")[sensor_columns].std()
hourly_std.columns = [
    f"{col}_std" for col in sensor_columns
]

hourly_min = X_hourly.groupby("hour")[sensor_columns].min()
hourly_min.columns = [
    f"{col}_min" for col in sensor_columns
]

hourly_max = X_hourly.groupby("hour")[sensor_columns].max()
hourly_max.columns = [
    f"{col}_max" for col in sensor_columns
]

hourly_features = pd.concat(
    [
        hourly_mean,
        hourly_std,
        hourly_min,
        hourly_max
    ],
    axis=1
)

hourly_features.reset_index(inplace=True)

print("Hourly feature dataset shape:")
print(hourly_features.shape)

display(hourly_features.head())
```

```python
Y_hourly = Y.copy()

Y_hourly["hour"] = Y_hourly["date_time"].dt.floor("h")

Y_hourly = Y_hourly[
    ["hour", "quality"]
]

display(Y_hourly.head())
```

```python
merged = pd.merge(
    hourly_features,
    Y_hourly,
    on="hour",
    how="inner"
)

print("Merged dataset shape:", merged.shape)

display(merged.head())
```

```python
correlation_with_quality = (
    merged
    .drop(columns=["hour"])
    .corr(numeric_only=True)["quality"]
    .drop("quality")
    .sort_values(
        key=lambda x: abs(x),
        ascending=False
    )
)

display(
    correlation_with_quality.to_frame(
        name="Correlation with Quality"
    ).round(4)
)
```

```python
top_features = correlation_with_quality.abs().sort_values(
    ascending=False
).head(5).index

print("Top 5 features associated with quality:")

for feature in top_features:
    print(feature)
```

```python
for feature in top_features:

    plt.figure(figsize=(8, 5))

    plt.scatter(
        merged[feature],
        merged["quality"],
        alpha=0.3
    )

    plt.xlabel(feature)
    plt.ylabel("Product Quality")
    plt.title(
        f"{feature} vs Product Quality"
    )

    plt.tight_layout()
    plt.show()
```

```python
correlation_results = []

features_for_test = [
    col for col in merged.columns
    if col not in ["hour", "quality"]
]

for feature in features_for_test:

    x = merged[feature]
    y = merged["quality"]

    mask = x.notna() & y.notna()

    pearson_r, pearson_p = stats.pearsonr(
        x[mask],
        y[mask]
    )

    spearman_r, spearman_p = stats.spearmanr(
        x[mask],
        y[mask]
    )

    correlation_results.append({
        "Feature": feature,
        "Pearson r": pearson_r,
        "Pearson p-value": pearson_p,
        "Spearman rho": spearman_r,
        "Spearman p-value": spearman_p
    })

correlation_results = pd.DataFrame(
    correlation_results
)

correlation_results["abs_Pearson"] = (
    correlation_results["Pearson r"].abs()
)

correlation_results = correlation_results.sort_values(
    "abs_Pearson",
    ascending=False
)

display(
    correlation_results.round(5)
)
```

```python
significant = correlation_results[
    correlation_results["Pearson p-value"] < 0.05
]

print(
    "Number of statistically significant features:",
    len(significant)
)

display(
    significant[
        [
            "Feature",
            "Pearson r",
            "Pearson p-value",
            "Spearman rho",
            "Spearman p-value"
        ]
    ].round(6)
)
```

```python
mean_features = [
    col for col in merged.columns
    if col.endswith("_mean")
]

corr_matrix = merged[
    mean_features + ["quality"]
].corr()

plt.figure(figsize=(14, 10))

sns.heatmap(
    corr_matrix,
    cmap="coolwarm",
    center=0,
    annot=False
)

plt.title(
    "Correlation Matrix of Process Variables and Product Quality"
)

plt.tight_layout()
plt.show()
```

```python
chamber_results = []

for chamber in range(1, 6):

    chamber_features = [
        col for col in mean_features
        if f"T_data_{chamber}_" in col
    ]

    correlations = merged[
        chamber_features + ["quality"]
    ].corr()["quality"].drop("quality")

    chamber_results.append({
        "Chamber": chamber,
        "Mean Absolute Correlation": correlations.abs().mean(),
        "Maximum Absolute Correlation": correlations.abs().max()
    })

chamber_df = pd.DataFrame(chamber_results)

display(
    chamber_df.sort_values(
        "Mean Absolute Correlation",
        ascending=False
    ).round(4)
)
```

```python
Y_plot = Y.sort_values("date_time").copy()

Y_plot["quality_rolling_mean"] = (
    Y_plot["quality"]
    .rolling(
        window=24,
        min_periods=1
    )
    .mean()
)

plt.figure(figsize=(16, 6))

plt.plot(
    Y_plot["date_time"],
    Y_plot["quality"],
    alpha=0.3,
    label="Observed Quality"
)

plt.plot(
    Y_plot["date_time"],
    Y_plot["quality_rolling_mean"],
    linewidth=2,
    label="24-hour Rolling Mean"
)

plt.xlabel("Time")
plt.ylabel("Product Quality")
plt.title("Product Quality and 24-Hour Rolling Mean")

plt.legend()

plt.tight_layout()
plt.show()
```

```python
Y_plot["quality_rolling_std"] = (
    Y_plot["quality"]
    .rolling(
        window=24,
        min_periods=1
    )
    .std()
)

plt.figure(figsize=(16, 6))

plt.plot(
    Y_plot["date_time"],
    Y_plot["quality_rolling_std"]
)

plt.xlabel("Time")
plt.ylabel("Rolling Standard Deviation")
plt.title("24-Hour Rolling Variability of Product Quality")

plt.tight_layout()
plt.show()
```

```python
def rolling_slope(series):

    x = np.arange(len(series))

    if len(series) < 2:
        return np.nan

    return np.polyfit(x, series, 1)[0]


Y_plot["quality_slope"] = (
    Y_plot["quality"]
    .rolling(window=24)
    .apply(rolling_slope, raw=False)
)

plt.figure(figsize=(16, 6))

plt.plot(
    Y_plot["date_time"],
    Y_plot["quality_slope"]
)

plt.axhline(
    0,
    linestyle="--"
)

plt.xlabel("Time")
plt.ylabel("Quality Trend Slope")
plt.title("Rolling Quality Trend")

plt.tight_layout()
plt.show()
```
