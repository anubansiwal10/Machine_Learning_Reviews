```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.cluster import KMeans
from sklearn.model_selection import GroupKFold
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.metrics import pairwise_distances
from scipy.spatial import Delaunay

# Load arsenic data
arsenic = pd.read_excel(
    os.path.join(path, "Contamination_Hotspots.xlsx"),
    sheet_name="Sheet4"
)

# Convert numerical columns
for col in ["As(ppb)", "Latitude (DD)", "Longitude (DD)"]:
    arsenic[col] = pd.to_numeric(arsenic[col], errors="coerce")

# Remove incomplete observations
arsenic.dropna(
    subset=["As(ppb)", "Latitude (DD)", "Longitude (DD)"],
    inplace=True
)

# Keep grid samples only
sample_col = "Type of Sample in grid (Identified hot spot/Grid Sample)"

grid = arsenic[
    arsenic[sample_col].str.strip().str.lower() == "grid sample"
].copy()

# Select Uttar Pradesh
up = grid[grid["State/UT"] == "Uttar Pradesh"].copy()

# Identify the district with the most grid observations
district_counts = up["District"].value_counts()

print("Uttar Pradesh district distribution:")
display(district_counts.to_frame("Observations"))

selected_district = district_counts.index[0]

print("\nSelected district:", selected_district)

data = up[up["District"] == selected_district].copy()

# Aggregate measurements at identical coordinates
data = (
    data.groupby(
        ["Latitude (DD)", "Longitude (DD)"],
        as_index=False
    )
    .agg({
        "As(ppb)": "mean",
        "District": "first",
        "State/UT": "first"
    })
)

print("\nFinal modelling dataset:", data.shape)
display(data.head())
```

```python

mean_lat = data["Latitude (DD)"].mean()

# Convert longitude and latitude differences into approximate kilometres
X = np.column_stack([
    data["Longitude (DD)"].values * 111.32 * np.cos(np.radians(mean_lat)),
    data["Latitude (DD)"].values * 110.574
])

y = data["As(ppb)"].values

print("Input shape:", X.shape)
print("Target shape:", y.shape)

print("Latitude range:",
      data["Latitude (DD)"].min(),
      "to",
      data["Latitude (DD)"].max())

print("Longitude range:",
      data["Longitude (DD)"].min(),
      "to",
      data["Longitude (DD)"].max())

print("Arsenic range:", y.min(), "to", y.max(), "ppb")
```

```python

class AdaptiveRBF:

    def __init__(self, n_centers=8, alpha=1.0, adaptive=True):
        self.n_centers = n_centers
        self.alpha = alpha
        self.adaptive = adaptive

    def _basis(self, X):
        distances = pairwise_distances(X, self.centers)

        return np.exp(
            -(distances ** 2) / (2 * self.sigmas[None, :] ** 2)
        )

    def fit(self, X, y):

        # 1. Select basis centres
        self.n_centers = min(self.n_centers, len(X))

        self.kmeans = KMeans(
            n_clusters=self.n_centers,
            random_state=42,
            n_init=10
        )

        labels = self.kmeans.fit_predict(X)
        self.centers = self.kmeans.cluster_centers_

        # Distances between basis centres
        center_distances = pairwise_distances(self.centers)
        np.fill_diagonal(center_distances, np.inf)

        nearest_center = center_distances.min(axis=1)

        # Global width for conventional RBF
        global_width = max(
            np.median(nearest_center) * 0.7,
            1e-6
        )

        # 2. Estimate basis widths
        self.sigmas = []

        for j in range(self.n_centers):

            cluster_points = X[labels == j]

            if len(cluster_points) > 1:
                distances = np.linalg.norm(
                    cluster_points - self.centers[j],
                    axis=1
                )
                local_width = np.median(distances) * 1.5
            else:
                local_width = nearest_center[j] * 0.5

            if self.adaptive:
                sigma = max(
                    local_width,
                    nearest_center[j] * 0.35,
                    global_width * 0.05
                )
            else:
                sigma = global_width

            self.sigmas.append(sigma)

        self.sigmas = np.asarray(self.sigmas)

        # 3. Construct Gaussian basis matrix
        Phi = self._basis(X)

        # Add intercept
        A = np.column_stack([
            np.ones(len(X)),
            Phi
        ])

        # 4. Regularised least-squares optimisation
        penalty = np.eye(A.shape[1]) * self.alpha
        penalty[0, 0] = 0  # Do not penalise intercept

        self.weights = np.linalg.solve(
            A.T @ A + penalty,
            A.T @ y
        )

        return self

    def predict(self, X):

        Phi = self._basis(X)

        A = np.column_stack([
            np.ones(len(X)),
            Phi
        ])

        return A @ self.weights
```

```python

n = len(X)

if n < 20:
    raise ValueError(
        "Insufficient unique grid locations for this experiment. "
        "Choose a district with more observations."
    )

n_centers = min(12, max(3, n // 5))

adaptive_model = AdaptiveRBF(
    n_centers=n_centers,
    alpha=1.0,
    adaptive=True
)

adaptive_model.fit(X, y)

predictions = adaptive_model.predict(X)

print("Adaptive RBF trained successfully!")
print("Number of basis centres:", adaptive_model.n_centers)
print("Number of adaptive widths:", len(adaptive_model.sigmas))
print("Learned weights:", adaptive_model.weights)

print("\nBasis widths:")
print(adaptive_model.sigmas)
```

```python

def idw_predict(X_train, y_train, X_test, power=2):

    distances = pairwise_distances(X_test, X_train)

    predictions = []

    for row in distances:

        # Handle an exact coordinate match
        exact = np.where(row < 1e-10)[0]

        if len(exact) > 0:
            predictions.append(np.mean(y_train[exact]))
            continue

        weights = 1 / np.maximum(row, 1e-10) ** power

        prediction = np.sum(weights * y_train) / np.sum(weights)

        predictions.append(prediction)

    return np.array(predictions)


# Create five geographical groups
spatial_clusters = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
).fit_predict(X)

cv = GroupKFold(n_splits=5)

results = {
    "Adaptive RBF": [],
    "Conventional RBF": [],
    "IDW": []
}

actual_values = []
predicted_values = {
    "Adaptive RBF": [],
    "Conventional RBF": [],
    "IDW": []
}

for fold, (train_idx, test_idx) in enumerate(
    cv.split(X, y, groups=spatial_clusters), start=1
):

    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    k = min(12, max(3, len(X_train) // 5))

    # Adaptive RBF
    adaptive = AdaptiveRBF(
        n_centers=k,
        alpha=1.0,
        adaptive=True
    )
    adaptive.fit(X_train, y_train)
    pred_adaptive = adaptive.predict(X_test)

    # Conventional RBF
    conventional = AdaptiveRBF(
        n_centers=k,
        alpha=1.0,
        adaptive=False
    )
    conventional.fit(X_train, y_train)
    pred_conventional = conventional.predict(X_test)

    # IDW
    pred_idw = idw_predict(
        X_train, y_train, X_test
    )

    predictions_fold = {
        "Adaptive RBF": pred_adaptive,
        "Conventional RBF": pred_conventional,
        "IDW": pred_idw
    }

    actual_values.extend(y_test)

    for name, pred in predictions_fold.items():
        predicted_values[name].extend(pred)

    print(f"Fold {fold} completed. "
          f"Train: {len(train_idx)}, Test: {len(test_idx)}")


# Calculate final out-of-fold metrics
actual_values = np.asarray(actual_values)

for name in results:

    pred = np.asarray(predicted_values[name])

    results[name] = {
        "MAE": mean_absolute_error(actual_values, pred),
        "RMSE": np.sqrt(mean_squared_error(actual_values, pred)),
        "R2": r2_score(actual_values, pred)
    }

results_df = pd.DataFrame(results).T

print("\nSPATIAL CROSS-VALIDATION RESULTS")
display(results_df.round(4))
```

```python

# Train final model on all available observations
final_model = AdaptiveRBF(
    n_centers=min(12, max(3, len(X) // 5)),
    alpha=1.0,
    adaptive=True
)

final_model.fit(X, y)

# Create geographical prediction grid
x_min, x_max = X[:, 0].min(), X[:, 0].max()
y_min, y_max = X[:, 1].min(), X[:, 1].max()

xx, yy = np.meshgrid(
    np.linspace(x_min, x_max, 120),
    np.linspace(y_min, y_max, 120)
)

grid_points = np.column_stack([
    xx.ravel(),
    yy.ravel()
])

# Restrict predictions to sampled convex hull
hull = Delaunay(X)

inside = hull.find_simplex(grid_points) >= 0

# Predict only inside the sampled region
predicted_grid = np.full(len(grid_points), np.nan)

predicted_grid[inside] = final_model.predict(
    grid_points[inside]
)

Z = predicted_grid.reshape(xx.shape)

# Convert coordinates back to longitude and latitude
lon_grid = xx / (111.32 * np.cos(np.radians(mean_lat)))
lat_grid = yy / 110.574

# Plot
plt.figure(figsize=(12, 8))

contour = plt.contourf(
    lon_grid,
    lat_grid,
    Z,
    levels=25,
    cmap="YlOrRd"
)

plt.scatter(
    data["Longitude (DD)"],
    data["Latitude (DD)"],
    c=data["As(ppb)"],
    cmap="YlOrRd",
    edgecolors="black",
    s=35,
    label="Observed grid samples"
)

plt.colorbar(
    contour,
    label="Predicted arsenic concentration (ppb)"
)

plt.xlabel("Longitude")
plt.ylabel("Latitude")

plt.title(
    f"Adaptive RBF Groundwater Arsenic Prediction\n"
    f"{selected_district}, Uttar Pradesh"
)

plt.legend()
plt.tight_layout()
plt.show()
```

```python

sampling_percentages = [10, 20, 30, 50, 70]

sparse_results = []

# Use a fixed random seed for reproducibility
rng = np.random.default_rng(42)

for percentage in sampling_percentages:

    n_samples = max(
        5,
        int(len(X) * percentage / 100)
    )

    if n_samples >= len(X):
        continue

    selected = rng.choice(
        len(X),
        size=n_samples,
        replace=False
    )

    X_subset = X[selected]
    y_subset = y[selected]

    model = AdaptiveRBF(
        n_centers=min(12, max(2, n_samples // 5)),
        alpha=1.0,
        adaptive=True
    )

    model.fit(X_subset, y_subset)

    # Evaluate on locations not selected for training
    remaining = np.setdiff1d(
        np.arange(len(X)),
        selected
    )

    predictions_sparse = model.predict(X[remaining])

    sparse_results.append({
        "Sampling_Percentage": percentage,
        "Training_Samples": n_samples,
        "MAE": mean_absolute_error(
            y[remaining], predictions_sparse
        ),
        "RMSE": np.sqrt(mean_squared_error(
            y[remaining], predictions_sparse
        )),
        "R2": r2_score(
            y[remaining], predictions_sparse
        )
    })

sparse_df = pd.DataFrame(sparse_results)

display(sparse_df.round(4))

plt.figure(figsize=(9, 5))

plt.plot(
    sparse_df["Sampling_Percentage"],
    sparse_df["RMSE"],
    marker="o"
)

plt.xlabel("Training Sampling Availability (%)")
plt.ylabel("RMSE (ppb)")
plt.title("Adaptive RBF Performance Under Limited Sampling")
plt.grid(alpha=0.3)
plt.show()
```
